```python
import numpy as np
from iterative_random_forest import IterativeRandomForest
from sklearn.datasets import make_classification

# Toy dataset
X, y = make_classification(n_samples=1000, n_features=20,
                           n_informative=10, random_state=42)

# Fit Iterative Random Forest
irf = IterativeRandomForest(n_estimators=100, n_iter=5, random_state=42)
irf.fit(X, y)

# Feature importances after last iteration
print("Feature importances:", irf.feature_importances_)

# Stable interactions
interactions = irf.get_stable_interactions()
print("Stable feature interactions:", interactions)
```
