```python
import pandas as pd
```

```python
#classification dataframe
models = ['RF', 'MLR', 'XGBoost', 'MLP', 'SVM-lin', 'SVM-RBF']

# Create a DataFrame with models as rows and accuracy metrics as columns
results_df = pd.DataFrame(index=models, columns = ["Accuracy"])

# Save to file
results_df.to_csv("cancer_baseline_results.csv")
```

```python
#classification minipatch dataframe
models = ['RF', 'Tree MP', 'XGBoost', 'Boosted MP', 'MLP', 'MLP MP']

# Create a DataFrame with models as rows and accuracy metrics as columns
mp_df = pd.DataFrame(index=models, columns = ["Accuracy"])

baseline_df = pd.read_csv("/Users/mariahloehr/IICD/IICD/Cancer Treatment/cancer_baseline_results.csv", index_col=0)

mp_df.loc["RF", "Accuracy"] = baseline_df.loc["RF", "Accuracy"]
mp_df.loc["XGBoost", "Accuracy"] = baseline_df.loc["XGBoost", "Accuracy"]
mp_df.loc["MLP", "Accuracy"] = baseline_df.loc["MLP", "Accuracy"]

# Save to file
mp_df.to_csv("cancer_mp_results.csv")
```
