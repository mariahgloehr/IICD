```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load data
results_df = pd.read_csv("/Users/mariahloehr/IICD/IICD/Cancer Treatment/cancer_baseline_results.csv", index_col=0)
```

```python
# Plot all bars
ax = results_df.plot(kind="bar", figsize=(9, 7), color="skyblue", legend=False)

for p in ax.patches:
    ax.annotate(
        f"{p.get_height():.3f}",        # round to 3 decimals
        (p.get_x() + p.get_width() / 2., p.get_height()),
        ha='center', va='bottom',
        fontsize=10, color="black", rotation=0, xytext=(0, 3),
        textcoords="offset points"
    )

# Style
plt.title("Model Accuracy")
plt.ylabel("Accuracy")
plt.xticks(rotation=45, ha='right')
plt.ylim(top = 1)
plt.tight_layout()
plt.show()
```
