```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load data
regression_df = pd.read_csv("/Users/mariahloehr/IICD/IICD/Bar Plot/regression_results.csv", index_col=0)
#regression_df = pd.DataFrame(regression_df)
```

```python
# 0.6 regression dataframe
models = ['RF', 'Decision Tree MP', 'XGBoost', 'Boosting MP', 'MLP', 'MLP MP']
phases = ['Overall', 'G0', 'G1', 'G2', 'M', 'S']

# Create a DataFrame with models as rows and RMSE as columns
results_df = pd.DataFrame(index=models, columns=phases)

results_df.loc['RF'] = regression_df.loc['RF (full)'].values
results_df.loc['XGBoost'] = regression_df.loc['XGBoost (full)'].values
results_df.loc['MLP'] = regression_df.loc['MLP (full)'].values

# Save to file
results_df.to_csv("minipatch_results.csv")
```

```python
results_df_5 = pd.read_csv("/Users/mariahloehr/IICD/IICD/Bar Plot/5_minipatch_results.csv", index_col=0)
```

```python
# Plot all bars
ax = results_df.plot(kind="bar", figsize=(12, 6), alpha=0.2)  # dim all bars

# Highlight "Overall" bars
overall_bars = ax.containers[0]  # "Overall" is usually the first container
for bar in overall_bars:
    bar.set_alpha(1.0)           # full opacity
    bar.set_color('tab:blue')    # standout color (optional)

# Style
plt.title("Model and 0.6 Minipatch RMSE by Phase", fontsize = 14)
plt.ylabel("RMSE")
plt.xticks(rotation=45, ha='right', fontsize = 14)
plt.legend(title="Phase")
plt.tight_layout()
plt.show()
```
