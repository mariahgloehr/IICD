```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load data
df = pd.read_csv("/Users/mariahloehr/IICD/IICD/Bar Plot/regression_results.csv", index_col=0)
results_df = df[~df.index.to_series().str.contains("top 40")]
```

```python
results_df
```

```python
# Plot all bars
ax = results_df.plot(kind="bar", figsize=(12, 6), alpha=0.2)  # dim all bars

# Highlight "Overall" bars
overall_bars = ax.containers[0]  # "Overall" is usually the first container
for bar in overall_bars:
    bar.set_alpha(1.0)           # full opacity
    bar.set_color('tab:blue')    # standout color (optional)

# Style
plt.title("Model RMSE by Phase")
plt.ylabel("RMSE")
plt.xticks(rotation=45, ha='right')
plt.legend(title="Phase")
plt.tight_layout()
plt.show()
```
